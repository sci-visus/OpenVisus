### Examples with Fixed Size

**Do you have limited memory in your workspace?**

No worries! We have a new feature that lets you query the data that will actually fit into your memory.

_Just pass a parameter `size` with desirable quantity while reading the data and it will return the data with best approximation to your memory. See Examples Below_

```python
import OpenVisus as ov
import re
import time
import matplotlib.pyplot as plt


dataset_url="http://atlantis.sci.utah.edu/mod_visus?dataset=nex-gddp-cmip6"
dataset = ov.LoadDataset(dataset_url)
size='100kb'
quality=10
```

```python
data = dataset.read(size=size)
print(data.shape)
plt.imshow(data,origin='lower')
```

```python
data = dataset.read(quality=quality)
print(data.shape)
plt.imshow(data,origin='lower')
```
